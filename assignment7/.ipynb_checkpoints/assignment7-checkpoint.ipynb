## 9) Make a histogram the distribution of the Earthquake magnitudes

https://pandas.pydata.org/pandas-docs/version/0.23/generated/pandas.DataFrame.hist.html

https://matplotlib.org/api/_as_gen/matplotlib.pyplot.hist.html

Do one subplot for the filtered and one for the unfiltered dataset. Use a Logarithmic scale. What sort of relationship do you see?

```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
fig, ax = plt.subplots(figsize = (10,10),nrows = 2)
fig.tight_layout(h_pad=5)
ax[0].hist(df['mag'],bins = np.arange(-1,8,0.1),log=True)
ax[0].set_xlabel('Earthquake Magnitudes')
ax[0].set_ylabel('Frequency (Log Scale)')
ax[0].set_title('Distribution of Earthquake Magnitudes on a Logarithmic Scale')

ax[1].hist(df_mag4['mag'],bins = np.arange(-1,8,0.1),log=True)
ax[1].set_xlabel('Earthquake Magnitudes (Above 4)')
ax[1].set_ylabel('Frequency (Log Scale)')
ax[1].set_title('Distribution of Earthquake Magnitudes (Above 4) on a Logarithmic Scale')
ax[1].set_xlim(4,8)
```

For the unfiltered dataset, I see a distribution that it bimodal and perhaps slightly right-skewed. The peaks are around a magnitude of 1.2 and a magnitude of 4.3. I notice that the frequencies are highest in the magnitude range of 0.2-2.5, while the frequencies are lower for values below 0 and values between 5 and 8.

In the filtered dataset, I see a distribution that is asymmetric, unimodal, and strongly right-skewed. The only peak is around a magnitude of 4.3. Most data points are on the left, and the tail of the distribution stretches to the right. The frequencies decline as the magnitude increases.


## 10) Visualize the locations of earthquakes by making a scatterplot of their latitude and longitude
Use a two-column subplot with both the filtered and unfiltered datasets. Color the points by magnitude. Make it pretty.

What difference do you note between the filtered and unfiltered datasets?

```python
fig, ax = plt.subplots(figsize = (18,6),ncols = 2)
splot0 = ax[0].scatter(df['longitude'],df['latitude'], c=df['mag'],s=4, vmin = -0.97, vmax = 8.2)
fig.colorbar(splot0,label= 'Magnitude')
splot1 = ax[1].scatter(df_mag4['longitude'],df_mag4['latitude'],c=df_mag4['mag'], s=4,vmin = -0.97, vmax = 8.2)
fig.colorbar(splot1,label= 'Magnitude')
ax[0].set_title('Magnitude of Earthquakes in the Unfiltered Dataset, \n1/31/2014 - 12/01/2014')
ax[0].set_xlabel('Longitude (°)')
ax[0].set_ylabel('Latitude (°)')
ax[1].set_title('Magnitude of Earthquakes in the Filtered Dataset \n(Above a Magnitude of 4), 1/31/2014 - 12/01/2014')
ax[1].set_xlabel('Longitude (°)')
ax[1].set_ylabel('Latitude (°)')
```

The filtered dataset scatterplot (right) is largely similar to the unfiltered dataset scatterplot (left), but the unfiltered dataset scatterplot shows earthquakes with magnitudes below 4 in the coordinate range -170° to -50° longitude and 18° to 75° latitude (North America, particularly contintental US, Alaska, and Hawaii) which are colored shades of blue. There are also a few earthquakes below a magnitude of 4 around the area of 5° latitude, 0° longitude (central Africa) and the area of 55° latitude, 0° longitude (central Europe) in the unfiltered dataset that are absent in the filtered dataset. The filtered dataset has fewer earthquakes in the area of -170° to -50° longitude and 18° to 75° latitude (North America) because most of the earthquakes in that area in the unfiltered dataset are below magnitude 4.

It seems likely that the dataset's earthquakes with magnitudes below 4 are almost exclusively in North America because the data comes from the U.S. Geological Survey, which likely focuses on U.S. earthquake data.

The vast majority of earthquakes in the unfiltered dataset appear to have magnitudes between 3 and 6. These points are mostly blue-green, teal, and green shades.

The vast majority of earthquakes in the filtered dataset appear to have magnitudes between 4 and 6. These points are mostly teal and green shades. It makes sense that most would be in this range because anything below a magnitude of 4 is filtered out.

I also notice that most observed earthquakes appear to be located along the coasts of continents.
