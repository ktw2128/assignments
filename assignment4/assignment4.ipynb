1. Write a function to convert temperature from kelvin to celsius
and celsius to kelvin

```python
# Part III: Functions
# 1
def c_to_k(temp):
    """ Function to convert temperature celsius to kelvin """
    return temp + 273.15

def k_to_c(temp):
    """ Function to convert temperature from kelvin to celsius """
    return temp - 273.15
```

2. Write a function to convert temperature to fahrenheit. Include an optional keyword argument to specify whether the input is in celsius or kelvin. Call your previously defined functions if necessary.

```python
# Part III: Functions
# 2
def convert_to_f(temp,unit='C'):
    """  Function to convert temperature to fahrenheit. 
        Include an optional keyword argument to specify whether the input is in celsius or kelvin."""
    if unit == "Celsius" or unit == "celsius" or unit == "C" or unit == "°C":
        fahrenheit = (temp * 9/5) + 32
        return fahrenheit
    elif unit == "Kelvin" or unit == "K" or unit == "k":
        fahrenheit = (k_to_c(temp) * 9/5) + 32
        return fahrenheit
    else:
        print("Please rewrite the unit type as a string, specifying Celsius or Kelvin")
```

```python
# Part III: Functions
#4
def temp_unit_conversion(temp,unit):
    """ function that converts from farenheit and uses a keyword argument to specify 
    whether you want the output in celsius or kelvin"""
    celsius = ((temp - 32) * 5/9)
    if unit == "Celsius" or unit == "celsius" or unit == "C" or unit == "°C":
        return celsius
    elif unit == "Kelvin" or unit == "K" or unit == "k":
        return c_to_k(celsius)
    else:
        print("Please rewrite the unit type as a string, specifying Celsius or Kelvin")
```

6. Write a function takes one argument (height in meters) and returns two arguments (feet and inches)
(Consult the tutorial on numbers if you are stuck on how to implement this.)

```python
# Part III: Functions
#6
def meters_to_feet(meters):
    """Function that takes in height in meters and returns height in feet and inches."""
    total_feet = meters * 3.28084
    feet = int(total_feet)
    inches = round((total_feet - feet) * 12)
    if inches == 12:
        feet += 1
        inches = 0
    return feet, inches
```

7) Verify that the “round trip” conversion from and back to meters is consistent. Check for 3 different values of height in meters

```python
# Part III: Functions
#7
def feet_to_meters(feet,inches):
    """Function that takes in height in feet and inches and returns height in meters."""
    if inches > 0: 
        additional = inches / 12
        total_feet = feet + additional 
    elif inches == 0:
        total_feet = feet
    meters = total_feet / 3.28084
    return round(meters,2)
```
